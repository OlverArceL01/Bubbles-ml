# Data Processing & Modeling

In this step, the data will be processed to train a semantic segmentation model and evaluate its performance.

## Imports

````python
# System
import os
from pathlib import Path

# Image Processing
from PIL import Image

# Data Manipulation
import pandas as pd
import numpy as np

# Visualization
import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator
from matplotlib.colors import ListedColormap
import seaborn as sns

# Machine Learning
from ultralytics import YOLO
````

## Data Processing 

YOLO uses a `dataset.yaml` file to define the dataset paths and expects a specific directory structure.

The current structure is:
```bash
data/raw/ 
├── train/ 
│ ├── train_images/ 
│ └── train_masks/ 
├── valid/ 
│ ├── valid_images/ 
│ └── valid_masks/ 
└── test/ 
  ├── test_images/ 
  └── test_masks/
```
Therefore, we will reorganize the dataset into the following structure:
```bash
data/processed/ 
├── images/ 
│ ├── train/ 
│ ├── valid/ 
│ └── test/ 
└── masks/ 
  ├── train/ 
  ├── valid/ 
  └── test/
```

The `dataset.yaml` file will define the dataset paths and class names:
```yaml
path: ../bubbles/data/processed
train: images/train
val: images/valid
test: images/test

masks_dir: masks

names:
  0: background
  1: bubble
```

### Mask Preprocessing
According to the [Semantic Segmentation with Ultralytics YOLO documentation](https://docs.ultralytics.com/tasks/semantic#:~:text=Pixels%20with%20value%20255%20are%20treated%20as%20%22ignore%22%20and%20excluded%20from%20loss%20computation.) pixels with a value of `255` are treated as ignore and excluded from loss computation.

The masks currently use pixel values of `0` and `255`. To represent the two classes cosistently, we will convert the masks to binary values: `0` for background and `1` for bubbles.

````python
for split in ["train", "valid", "test"]:
    source_dir = Path("../../data/raw") / split / f"{split}_masks"
    target_dir = Path("../../data/processed/masks") / split
    target_dir.mkdir(parents=True, exist_ok=True)

    for path in source_dir.glob("*.png"):
        mask = np.array(Image.open(path).convert("L"))
        # Convert mask pixel values to binary (0 or 1)
        mask = (mask > 127).astype(np.uint8)
        Image.fromarray(mask).save(target_dir / path.name)

    print(f"{split}: {len(list(target_dir.glob('*.png')))} masks saved")
````

## Model Training on Local

The `YOLO26n-sem` model will be used, it is the smallest model in terms of parameter count and is designed for fast inference, making it a suitable candidate for fast training and inference.

Initially, the model will be trained on 10% of the approximately 13.000 training images for 5 epochs.

The results will be saved in `../../runs/semantic/experiment_local` and the best model weights will be saved in `../../runs/semantic/experiment_local/weights/best.pt`.

````python
# Load model
model = YOLO("yolo26n-sem.pt")

# Train model
results = model.train(
    data="../../dataset.yaml",
    epochs=5,
    imgsz=640,
    batch=4,
    fraction=0.1,
    seed=42,
    name="experiment_local"
)
````

### Evaluation

The model will be evaluated using the best weights obtained during training and the validation set.

Some metrics are not saved as expected, so the existing evaluation results are used to save them and calculate additional metrics.

The results will be saved in `../../runs/semantic/validation-local`.

````python
model = YOLO("../../runs/semantic/experiment_local/weights/best.pt")
metrics = model.val(name="validation-local")

def save_confusion_matrix(path):
    path = Path(path)
    cm = metrics.confusion_matrix.matrix
    cm_normalized = cm / (cm.sum(axis=0, keepdims=True) + 1e-9)
    df_cm = pd.DataFrame(cm)
    df_cm.to_csv( path / "confusion_matrix.csv", index=False)

def save_class_metrics(path):
    path = Path(path)
    df_cm = pd.read_csv( path / "confusion_matrix.csv")
    df_cm.columns = df_cm.columns.astype(int)
    TP = df_cm.loc[1,1]
    TN = df_cm.loc[0,0]
    FP = df_cm.loc[1,0]
    FN = df_cm.loc[0,1]
    bubble_precision = TP / (TP + FP)
    bubble_recall = TP / (TP + FN)
    bubble_iou = TP / (TP + FP + FN)
    bubble_dsc = 2 * TP / (FP + 2 * TP + FN)

    background_precision = TN / (TN + FN)
    background_recall = TN / (TN + FP)
    background_iou = TN / (TN + FP +  FN)
    background_dsc = 2 * TN / (FP + 2 * TN + FN)

    class_metrics = pd.DataFrame(
        columns = ["Class", "Precision", "Recall", "IoU", "DSC"],
        data = [
            ["background", background_precision, background_recall, background_iou, background_dsc],
            ["bubble", bubble_precision, bubble_recall, bubble_iou, bubble_dsc]
        ]
    )
    class_metrics.to_csv(path / "class_metrics.csv", index=False)


Path("../../metrics/experiment_local").mkdir(parents=True, exist_ok=True)
save_confusion_matrix("../../metrics/experiment_local")
save_class_metrics("../../metrics/experiment_local")
````

### Metrics plotting Functions
Functions for visualizing evaluation metrics.

````python
def plot_confusion_matrix(df_cm, save_path=None):
    labels = ["background", "bubble"]
    fig, ax = plt.subplots(figsize=(3, 3))
    sns.heatmap(
        df_cm,
        ax=ax,
        annot=True,
        fmt=".0f",
        cmap="Blues",
        xticklabels=labels,
        yticklabels=labels
    )
    ax.set_title("Confusion Matrix")
    ax.set_xlabel("True")
    ax.set_ylabel("Predicted")
    if save_path:
        plt.savefig(save_path, dpi=300, bbox_inches="tight")
    plt.show()

def plot_metrics_epoch(results_df, save_path=None):
    fig, ((ax1, ax2, ax3, ax4)) = plt.subplots(1, 4, figsize=(20, 5))
    
    ax1.plot(results_df["epoch"], results_df["train/ce_loss"], color="blue", label="train")
    ax1.plot(results_df["epoch"], results_df["val/ce_loss"], color="orange", label="val")
    ax1.set_title("CE Loss")
    ax1.set_xlabel("Epoch")
    ax1.set_ylabel("Loss")
    ax1.legend()
    
    ax2.plot(results_df["epoch"], results_df["train/dice_loss"], color="blue", label="train")
    ax2.plot(results_df["epoch"], results_df["val/dice_loss"], color="orange", label="val")
    ax2.set_title("DICE Loss")
    ax2.set_xlabel("Epoch")
    ax2.set_ylabel("Loss")
    ax2.legend()
    
    ax3.plot(results_df["epoch"], results_df["metrics/mIoU"], color="blue")
    ax3.set_title("mIoU")
    ax3.set_xlabel("Epoch")
    ax3.set_ylabel("mIoU")
    
    ax4.plot(results_df["epoch"], results_df["metrics/pixel_acc"], color="blue")
    ax4.set_title("Pixel Accuracy")
    ax4.set_xlabel("Epoch")
    ax4.set_ylabel("Pixel Accuracy")
    
    for ax in [ax1, ax2, ax3, ax4]:
        ax.xaxis.set_major_locator(MaxNLocator(integer=True))

    if save_path:
        plt.savefig(save_path, dpi=300, bbox_inches="tight")
    plt.show()
````

### Confusion Matrix
The confusion matrix highlights the class imbalance between background and bubble pixels.

````python
df_cm = pd.read_csv("../../metrics/experiment_local/confusion_matrix.csv")
plot_confusion_matrix(df_cm, "../../images/experiment_local_confusion_matrix.png")
````

### Class Metrics
The bubble class shows room for improvement in both precision and recall. With a precision of 0.57, the model frequently misclassified background pixels as bubbles, while a recall of 0.67 indicates that it identifies 67% of the actual bubble pixels.

The bubble IoU of 0.45 and Dice Similarity Coefficient of 0.62 indicate limited overlap between the predicted and ground truth masks, highlighting the model's difficulty in accuractely segmenting bubbles.

````python
class_metrics = pd.read_csv("../../metrics/experiment_local/class_metrics.csv")
print(class_metrics)
````

### Metrics vs Epoch
Both Cross-Entropy and Dice loss decrease rapidly during the first two epochs, reaching values between 0.2 and 0.3. No clear signs of severe overfitting are observed within these five epochs.


Mean IoU increases rapidly during the first two epochs and then stabilizes around 0.67. Pixel Accuracy follows a similar trends, stabilizing around 0.89.

````python
results_df = pd.read_csv("../../runs/semantic/experiment_local/results.csv")
plot_metrics_epoch(results_df, "../../images/experiment_local_metrics_vs_epoch.png")
````

### Predictions

Five images will be randomly selected to compare the original images, ground-truth masks, and predicted masks generated by the previously trained model.

It can be observed that the ground-truth masks tend to label bubbles outside the froth surface, which could be considered annotation noise.

````python
np.random.seed(42)
model = YOLO("../../runs/semantic/experiment_local/weights/best.pt")

def make_predictions(model, split, size, figsize, save_path=None):
    names = np.random.choice([f for f in os.listdir(Path("../../data/processed/images") / split) if f.endswith(".png")], size=size, replace=False)
    images = [np.array(Image.open(Path("../../data/processed/images") / split / name)) for name in names]
    masks = [np.array(Image.open(Path("../../data/processed/masks") / split / name)) for name in names]
    
    results = model.predict(source=images, imgsz=640, verbose=False)
    preds = [result.semantic_mask.data.cpu().numpy() for result in results]
    
    fig, axes = plt.subplots(size, 3, figsize=figsize)
    cmap = ListedColormap(["blue", "red"])
    
    for i in range(size):
        mask_bubble_pixels = np.count_nonzero(masks[i] == 1)
        mask_bubble_area_percentage = mask_bubble_pixels / masks[i].size * 100

        pred_bubble_pixels = np.count_nonzero(preds[i] == 1)
        pred_bubble_area_percentage = pred_bubble_pixels / preds[i].size * 100

        
        axes[i, 0].imshow(images[i])
        axes[i, 0].set_title("Original")
        axes[i, 0].axis("off")
        
        axes[i, 1].imshow(images[i])
        axes[i, 1].imshow(masks[i], alpha=0.5, cmap="bwr")
        axes[i, 1].contour(masks[i], levels=[0.5], colors=["red"], linewidths=1)
        axes[i, 1].set_title(f"Ground Truth - BAP: {mask_bubble_area_percentage:.2f}%")
        axes[i, 1].axis("off")
        
        axes[i, 2].imshow(images[i])
        axes[i, 2].imshow(preds[i], alpha=0.5, cmap="bwr")
        axes[i, 2].contour(preds[i], levels=[0.5], colors=["red"], linewidths=1)
        axes[i, 2].set_title(f"Prediction - BAP: {pred_bubble_area_percentage:.2f}%")        
        axes[i, 2].axis("off")
    plt.tight_layout()
    if save_path:
        plt.savefig(save_path, dpi=300, bbox_inches="tight")
    plt.show()
make_predictions(model, "valid", 5, (15,15), "../../images/experiment_local_predictions.png")
````

## Model Training on Google Cloud

A `yolo26n-sem` model will be trained on Google Cloud using the full training dataset for 50 epochs. A training script, `src/bubbles/train.py` will also be created.

The images and masks from `data/processed` will be compressed into an archive and uploaded to a Google Cloud Storage bucket.

A Compute Engine Virtual Machine with 4 vCPUs, 15 GB of RAM, and NVIDIA L4 GPU, and a 50 GB balanced persistent disk will be used for training. The dataset archive will then be copied form the bucket and extracted on the VM.

Finally, training will be launched using `uv run train --model=nano`. tmux will be used to keep the training session running after disconnecting from SSH.
