# Data Collection
Froth Flotation is critical in mineral processing and metallurgical applications, which is an efficient method of beneficiation of minerals.
Froth Flotation involves passing air bubbles through a mixture of ore particles and water, separing valuable components from a mixture.

The dataset used is provided by Daria Ivanycheva, a student at ITMO.

## Data Source
Accessed on October 2026
- Froth bubbles segmentation dataset: https://www.kaggle.com/datasets/obobojk/froth-bubbles/data

The dataset contains 20.000 segmented images, frames of 14 flotation videos that have been segmented using Yolov8, segment anything models and morphological based methods.

Each image is 800x600 pixels and saved in `.png` format.

## Imports

```python
import os
from PIL import Image
from pathlib import Path
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
```

## Dataset Size
The number of images and masks in the train, validation, and test sets is printed below.

```python
train_images = os.listdir("../../data/raw/train/train_images")
train_masks = os.listdir("../../data/raw/train/train_masks")
valid_images = os.listdir("../../data/raw/valid/valid_images")
valid_masks = os.listdir("../../data/raw/valid/valid_masks")
test_masks = os.listdir("../../data/raw/test/test_masks")
test_images = os.listdir("../../data/raw/test/test_images")

print("Train images:", len(train_images))
print("Train masks:", len(train_masks))
print("Valid images:", len(valid_images))
print("Valid masks:", len(valid_masks))
print("Test images:", len(test_images))
print("Test masks:", len(test_masks))
```

## Images and Masks
Next, some examples of images and their corresponding masks are shown.

```python
fig, axes = plt.subplots(4, 2, figsize=(20, 20))
cmap = ListedColormap(["blue", "red"])

for i in range(4):
    image = Image.open(Path("../../data/raw/train/train_images") / train_images[i]).convert("RGB")
    mask = Image.open(Path("../../data/raw/train/train_masks") / train_masks[i])
    axes[i, 0].imshow(image)
    axes[i, 0].set_title(f"Image {i + 1}")
    axes[i, 0].axis("off")

    axes[i, 1].imshow(image)
    axes[i, 1].imshow(mask, alpha=0.5, cmap="bwr")
    axes[i, 1].contour(mask, levels=[0.5], colors=["red"], linewidths=1)
    axes[i, 1].set_title(f"Mask {i + 1}")
    axes[i, 1].axis("off")

plt.tight_layout()
plt.show()
```
